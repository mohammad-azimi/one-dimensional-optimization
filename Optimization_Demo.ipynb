## Check the error against the reference

برای مثال اصلی جواب دقیق معلوم است. برای تابع ناشناخته از این بخش صرف‌نظر
می‌کنیم و بازه‌ی نهایی و فرض یونیمودال بودن را بررسی می‌کنیم.


```python
if problem.REFERENCE_X is None:
    print('No exact reference supplied for this objective.')
else:
    for row in payload['results']:
        error = abs(row['x'] - problem.REFERENCE_X)
        passed = error <= row['delta'] * (1 + 1e-10)
        print(f"{row['method']:<16} delta={row['delta']:g} "
              f"error={error:.2e}  {'PASS' if passed else 'FAIL'}")
```
